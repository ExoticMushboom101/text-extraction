```python
import pandas as pd

# Sample data for the CSV
data = {
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age": [30, 25, 35, 40, 29],
    "Salary": [70000, 48000, 85000, 95000, 52000],
    "Department": ["Engineering", "HR", "Marketing", "Engineering", "HR"]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save as CSV
csv_file_name = "sample_data.csv"
df.to_csv(csv_file_name, index=False)

print(f"CSV file '{csv_file_name}' created successfully!")
```
